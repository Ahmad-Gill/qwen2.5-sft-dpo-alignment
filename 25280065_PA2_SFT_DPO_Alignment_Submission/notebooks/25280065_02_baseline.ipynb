```
import random, torch
from huggingface_hub import hf_hub_download
from transformers import AutoTokenizer, AutoModelForCausalLM

DEBUG = False          # True = quick test (3 questions, short answers)
MODEL = "Qwen/Qwen2.5-0.5B"
PROMPT = "### Instruction:\n{}\n\n### Response:\n"
GEN = dict(do_sample=False, max_new_tokens=32 if DEBUG else 256, repetition_penalty=1.1)


def get_questions():
    path = hf_hub_download("tatsu-lab/alpaca_eval", "alpaca_eval.json", repo_type="dataset")
    data = json.load(open(path))
    picked = sorted(random.Random(42).sample(range(len(data)), 300))
    return [{"id": f"alpaca_eval_{i}", "instruction": data[i]["instruction"]} for i in picked]


@torch.no_grad()
def answer(tok, model, questions, batch_size=16):

    results = []
    for i in range(0, len(questions), batch_size):
        batch = questions[i:i + batch_size]
        enc = tok([PROMPT.format(q["instruction"]) for q in batch],
                  return_tensors="pt", padding=True).to(model.device)
        out = model.generate(**enc, **GEN, pad_token_id=tok.pad_token_id)
        texts = tok.batch_decode(out[:, enc["input_ids"].shape[1]:], skip_special_tokens=True)
        for q, text in zip(batch, texts):
            results.append({**q, "output": text.split("### Instruction")[0].strip(),
                            "generator": "qwen2.5-0.5b-base", "dataset": "alpaca_eval"})
        print(f"{len(results)}/{len(questions)} done")
    return results
```
